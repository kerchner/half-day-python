# Iteration (looping)

Iteration allows us to repeat over a section of code, and iterate through a list (or other "iterable") at the same time.  **`for`** and **`while`** create iterations, like this:

```
numbers = [4, 6, 0, 5.5, 3]
for n in numbers:
    print(f"The next number in the list, squared is {n**2}")
    
print("We're done iterating -- notice that this line isn't indented, so it's outside the loop")
```    

and this brings up the topic of **indentation**!  The block (i.e., lines) of code that you're iterating over needs to be indented.  In Python, you should indent by 4 spaces.

Try creating a list of exam scores that we'd like to grade on a curve.  Use `max()` to find out the highest score.  Then create an iteration that prints out the score as well as the score graded on a curve (by dividing by the highest score).


````python
scores = [70, 90, 81, 93, 77, 86, 85]

print(max(scores))
````

````python
for s in scores:
    # The "round()" function rounds to 'ndigits' number of decimal places
    curved_score = round(100 * s / max(scores), ndigits=1)
    print(f"The score was {s} which is {curved_score} when curved.")
````

Iterating over dictionaries is slightly different:

### A bit more about passing values to functions

Remember earlier how we graded test scores on a curve?  Let's make a function to do that.  We'll pass it the individual test score, and the top score that will be our new "100%":

````python
def curve(score, top_score):
    return 100*score/top_score
````

````python
curve(89, 96)
````

Now, if we specify which parameter is which, we can pass them in any order we like!

````python
curve(top_score=96, score=89)
````

Key points here:

- definition starts with **`def`**
- function body is indented
- parameters are matched "in order", unless we specify them by name
- **`return`** keyword precedes returned value

# Challenge - Word Counter

We would like to create a function called `wordcount` that accepts a string of text, and returns a dictionary with the number of times each word appears.

For example:

`wordcount('This sentence is a sentence that repeats itself many many times')`

should return:

```
{'This': 1,
 'sentence': 2,
 'is': 1,
 'a': 1,
 'that': 1,
 'repeats': 1,
 'itself': 1,
 'many': 2,
 'times': 1}
 ```

 Hint: If `s` is a string variable, you can use `s.split(' ')` to break s into a list of words, where Python breaks the sentence up on spaces (or on whatever you pass to `.split()`

````python
def wordcount(sentence):
  word_dict = {}
  words = sentence.split(' ')
  for word in words:
    if word in word_dict:
      word_dict[word] = word_dict[word] + 1
    else:
      word_dict[word] = 1
  return word_dict
````

````python
wordcount('This is a sentence with some words and some more words that have repeated words')
````

We've managed to essentially build our own version of what `Counter` from the `collections` library does!

Can we improve our `wordcount` function by adding an optional parameter `case_sensitive` that makes it case-sensitive or not?  Let's say that the default should be for it to be case-sensitive.  We can implement this by converting all words to lower case (when `case_sensitive` is `True`).  The function to do this is called `lower()`, so:

```
text = 'The cat jumped over Michael'
text.lower()
```
would yield:

`the cat jumped over michael`

````python
def wordcount(sentence, case_sensitive=True):
    if case_sensitive:
        sentence = sentence.lower()
    words = sentence.split(" ")
    word_dict = {} # create an empty dictionary
    for word in words:
        if word in word_dict:
            word_dict[word] = word_dict[word] + 1
        else:
            word_dict[word] = 1
    return word_dict
````

# Challenge - Putting some skills together

Can you create a function called `pad` that would take a list and pads it with some new item if it's less than the length you want?  You would use it like this:

```
names = ['Ali', 'Bob', 'Carla']
names = pad(names, 5, '*')
```
Then `names` would evaluate to:
```
['Ali', 'Bob', 'Carla', '*', '*']
```

````python
def pad(list_in, desired_length, pad_value):
  pad_length = desired_length - len(list_in)  # Warning: We're assuming desired_length > len(list_in) which might not always be true
  list_out = list_in + [pad_value] * pad_length

  return list_out
````

````python
pad(['x', 'y'], 8, 'XOXO')
````
